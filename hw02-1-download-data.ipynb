# HW02 Part 1: Download Wikipedia Data

This notebook downloads the hourly Wikipedia event data from the public S3 bucket to a local `data` directory on EC2.

```python
import boto3
from pathlib import Path
```

## Set up the S3 source and local data directory

```python
bucket_name = "dsan6000-wikipedia"
prefix = "hourly_parquet/"

data_dir = Path("data")
data_dir.mkdir(exist_ok=True)
```

## List the hourly Parquet files in S3

This section connects to the public S3 bucket and identifies the hourly Parquet files that need to be downloaded.

```python
s3 = boto3.client("s3")

response = s3.list_objects_v2(
    Bucket=bucket_name,
    Prefix=prefix
)

parquet_files = [
    obj["Key"]
    for obj in response.get("Contents", [])
    if obj["Key"].endswith(".parquet")
]

len(parquet_files), parquet_files[:5]
```

## Download the Parquet files to EC2

Download each hourly Parquet file from the public S3 bucket to the local `data` directory.

```python
for key in parquet_files:
    filename = Path(key).name
    local_path = data_dir / filename

    s3.download_file(
        bucket_name,
        key,
        str(local_path)
    )

print(f"Downloaded {len(parquet_files)} parquet files to {data_dir}/")
```

## Verify the downloaded files

Check that the Parquet files were successfully downloaded to the local `data` directory.

```python
downloaded_files = sorted(data_dir.glob("*.parquet"))

print(f"Number of downloaded files: {len(downloaded_files)}")
downloaded_files[:5]
```
